```python
from groq import Groq

client = Groq(api_key="XYZ")


def generate_player(persona_input):
    prompt = f"""
You are a professional cricket analyst.

Generate a realistic cricket player persona based on the input below.

Input: {persona_input}

Strict Instructions:
- The player must sound like a real international or domestic cricketer
- Do NOT mention that the player is fictional
- Keep details believable and practical
- Avoid repeating common famous player names
- Use clean and structured format

Output Format:

Name:
Age:
Country:
Role:
Batting Style:
Bowling Style:
Strengths:
Weaknesses:
Career Highlights:
"""

    try:
        response = client.chat.completions.create(
            model="llama-3.1-8b-instant",
            messages=[{"role": "user", "content": prompt}],
            temperature=0.8   # adds creativity
        )

        return response.choices[0].message.content

    except Exception as e:
        return f"Error: {str(e)}"



if __name__ == "__main__":
    print("🏏 Cricket Player Persona Generator")
    print("-----------------------------------")

    while True:
        user_input = input("\nEnter player type (or type 'exit' to quit): ")

        if user_input.lower() == "exit":
            print("Goodbye! 👋")
            break

        result = generate_player(user_input)
        print("\nGenerated Player Persona:\n")
        print(result)
```
